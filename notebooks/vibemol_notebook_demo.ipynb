# VibeMol Notebook Demo

This notebook demonstrates two notebook workflows provided by `api.vibemol_client`:

1. `render_to_png(...)` for static image rendering.
2. `vibemol(...)` for embedding VibeMol in an iframe and auto-loading files.

Prerequisites (one-time):

```bash
python3 -m venv .venv
source .venv/bin/activate
pip install -r api/requirements.txt
python -m playwright install chromium
```

````python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    for p in [start, *start.parents]:
        if (p / "index.html").exists() and (p / "assets" / "data" / "sample.cube").exists():
            return p
    raise FileNotFoundError("Could not locate VibeMol project root from current working directory.")

PROJECT_ROOT = find_project_root(Path.cwd().resolve())
SAMPLE_CUBE = PROJECT_ROOT / "assets" / "data" / "sample.cube"
OUT_DIR = PROJECT_ROOT / "out" / "notebook"
OUT_DIR.mkdir(parents=True, exist_ok=True)
OUT_PNG = OUT_DIR / "sample_cube_notebook.png"

print("Project root:", PROJECT_ROOT)
print("Input cube:", SAMPLE_CUBE)
print("Output image:", OUT_PNG)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
````

````python
from api.vibemol_client import render_to_png, vibemol
````

## 1) Render to PNG (Playwright automation)

````python
render_to_png(
    SAMPLE_CUBE,
    OUT_PNG,
    style="toon",
    iso=0.02,
    wait_ms=900,
)
print(f"Saved: {OUT_PNG}")
````

````python
from IPython.display import Image, display
display(Image(filename=str(OUT_PNG)))
````

## 2) Embed VibeMol and auto-open files

### Hosted web app

````python
vibemol(SAMPLE_CUBE, source="web", height=760)
````

### Local app

Run this in a terminal (repo root):

```bash
python3 -m http.server 8000
```

Then run:

````python
vibemol(SAMPLE_CUBE, source="local", local_port=8000, height=760)
````
